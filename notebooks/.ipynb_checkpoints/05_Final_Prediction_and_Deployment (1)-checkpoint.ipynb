###  Crop Recommendation System — Final Prediction & Deployment
In this notebook, we’ll use the trained model to predict the best crop for given soil and weather conditions.

````python
# Step 1: Import necessary libraries
import pandas as pd
import numpy as np
import joblib

# Load preprocessing tools and trained model
model = joblib.load('../models/crop_model.pkl')
scaler = joblib.load('../models/scaler.pkl')
label_encoder = joblib.load('../models/label_encoder.pkl')

print(" Model, Scaler, and Label Encoder loaded successfully!")
````

````python
#  Step 2: Define prediction function

def recommend_crop(N, P, K, temperature, humidity, ph, rainfall):
    # Create a DataFrame for input values
    input_data = pd.DataFrame([[N, P, K, temperature, humidity, ph, rainfall]],
                              columns=['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall'])
    
    # Scale input
    input_scaled = scaler.transform(input_data)
    
    # Predict crop
    prediction = model.predict(input_scaled)
    crop_name = label_encoder.inverse_transform(prediction)[0]
    
    return crop_name
````

````python
# Step 3: Example prediction

# Example input (can be replaced with IoT sensor readings or manual values)
sample_input = {
    'N': 90,       # Nitrogen
    'P': 42,       # Phosphorous
    'K': 43,       # Potassium
    'temperature': 25.5,  # °C
    'humidity': 80,       # %
    'ph': 6.5,
    'rainfall': 200       # mm
}

predicted_crop = recommend_crop(**sample_input)
print(f" Recommended Crop: **{predicted_crop}**")
````

---
## Step 4: Deployment Ideas

You can deploy this model using:
- **Flask or FastAPI** (for a web app)
- **Streamlit or Gradio** (for an interactive dashboard)
- **IoT integration** (using Raspberry Pi + sensors)

Example (Flask):
```python
from flask import Flask, request, jsonify
app = Flask(__name__)

@app.route('/predict', methods=['POST'])
def predict():
    data = request.get_json()
    crop = recommend_crop(**data)
    return jsonify({'Recommended Crop': crop})
```
Run your app: `flask run`
